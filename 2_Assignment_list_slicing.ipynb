# Part A: Slicing Basics

```python
l = [10, 20, 30, 40, 50, 60, 70, 80]
print(l[2:5])
print(l[:4])
print(l[5:])
print(l[::2])
print(l[::-1])
print(l[-4:-1])
print(l[6:1:-2])
print(l[-3:])
```

# Part B: Nested Lists

```python
nested = [1, 2, [100, 200, [300, 400]], 5]
nested[2]
nested[2][1]
nested[2][2][1]
nested[2][1] = 250
print(nested)
```
