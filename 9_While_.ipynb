# While

```python
total = 0
num = 1
while num < 6:
    total = total + num
    num = num + 1
    print(total)
```

```python
age = 25

while age >= 18:
    print(age)
    age = age - 1
```

```python
couter = 10
while couter >= 0:
    print(couter)
    couter = couter - 1
```

```python
button = 'hi'

while button != 'exit':
    print(button)
    button = input("Please enter 'exit' to stop loop ")
```

```python
# Break, Continue, Pass
```

```python
for i in range(1,6):
    if i == 3:
        break
    print(i)
```

```python
for i in 'python':
    if i == 'h':
        break
    print(i)
```

# Continue



```python
for i in range(1,6):
    if i == 3:
        continue
    print(i)
    
    
```

```python
for i in 'hello':
    if i == 'l':
        continue
    print(i)
    
        
    
```

# Pass

```python
for i in range(1,6):
    if i == 3:
        pass
    else:
        print(i)
```
